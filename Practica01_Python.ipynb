Primera  parte de la práctica - Numpy y Matplotlip

```python
import numpy as np
import matplotlib.pyplot as plt

# a) Creación de vectores

a = np.array([3.1, 1, -0.5, -3.2, 6])
b = np.array([1, 3, 2.2, 5.1, 1])

print("Vector a:", a)
print("Vector b:", b)

# b) Producto escalar, se debe considera las dimensiones de los vectores deben ser igual para realizar la operación

producto_Escalar = np.dot(a, b)
print("Producto escalar:", producto_Escalar)

# c) Producto punto a punto

producto_Punto = a * b
print("Producto punto a punto:", producto_Punto)

# d) Construcción de una matriz A

A = np.array([
    [2,   -1,   -3],
    [4,    1.5, -2.5],
    [7.3, -0.9,  0.2]
])

print("Matriz A:\n", A)

# e) Transpuesta de la matriz A
transpuesta= A.T

print("Transpuesta de A:\n", transpuesta)

# f) Submatrices

print("Matriz de unos:\n", np.ones(A.shape)) #crea una matriz de unos con las misma dimensipon de la Matriz A
print("Round:\n", np.round(A, 1)) #Redondea los valores de LA MATRIZ A a 1 decimal
print("Ceil:\n", np.ceil(A)) #Redondea hacia arriba
print("Floor:\n", np.floor(A))#Redondea hacia Abajo

# g) Primera fila y  tercera columna de la matriz A
print("El valor es:", A[0,2])
# h) Segunda fila de la Matriz A
print("Segunda fila:", A[1,:])
# i) Dimensión de A
print("Dimensión de la matrz A:", A.shape)

"""GRAFICANDO - señales """
# Construcción de las funciones(señal)
n = np.arange(0,101)
y = np.sin(np.pi * 0.12 * n)
y2 = np.cos(2 * np.pi * 0.03 * n)

# Suma y prodcuto de las señales
suma = y + y2
producto= y * y2

# graficando las señales y y y2
#plt.figure()
fig =plt.figure(figsize=(10, 5))
plt.plot(n, y, label="y[n]")
plt.plot(n, y2, label="y2[n]")
plt.title("Señales y[n] y y2[n]")
plt.xlabel("n")
plt.ylabel("Amplitud")
plt.legend()
plt.grid()
plt.show()

# Graficando las señales (suma y producto )
fig =plt.figure(figsize=(10, 5))
plt.plot(n, suma, label="suma = y + y2")
plt.plot(n, producto, label="producto= y*y2")
plt.title("Señales Suma y Producto")
plt.xlabel("n")
plt.ylabel("Amplitud")
plt.legend()
plt.grid()
plt.show()
```

```python
import pandas as pd

def estadisticas_notas(diccionario_notas):
    serie = pd.Series(diccionario_notas)

    resultados = pd.Series({
        "Nota mínima": serie.min(),
        "Nota máxima": serie.max(),
        "Media": serie.mean(),
        "Desviación típica": serie.std()
    })
    return resultados
# Ejemplo con notas aleatorios de 5 estudiantes
notas = {
    "Juan": 4.5,
    "Ana": 3.8,
    "Luis": 4.9,
    "Marta": 2.9,
    "Pedro": 3.5
}
print(estadisticas_notas(notas))
```

```python
import pandas as pd
#cargando datos del archivo csv
df = pd.read_csv("datos.csv", sep=";") #el sep separa columnas del archvivo
#primera y ultimas filas
print("Primeras filas:", df.head())
print("\nÚltimas filas:", df.tail())
#Borramos columna unnamed: 0
df = df.drop(columns=["Unnamed: 0"])
# calculando IMC
df["BMI"] = df["Weight"] / ((df["Height"] / 100) ** 2)
#Categorizamos
def categoria_bmi(bmi):
    if bmi < 18.5:
        return "Bajo peso"
    elif 18.5 <= bmi < 25:
        return "Normal"
    elif 25 <= bmi < 30:
        return "Sobrepeso"
    else:
        return "Obesidad"

df["Categoria"] = df["BMI"].apply(categoria_bmi)

print(df.head())
```
